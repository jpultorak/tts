```python
from phonemizer import phonemize
from phonemizer.backend import EspeakBackend
from phonemizer.separator import Separator
import os

os.environ["PHONEMIZER_ESPEAK_LIBRARY"] = r"C:\Program Files\eSpeak NG\libespeak-ng.dll"
```

```python
s1 = "Hello world!aaa"
s2 = "Where the hood at?"
s3 = "Existence is like a disco, a dance of light and shadow, of being and not being."
texts = [s1, s2, s3]
```

```python
phn = phonemize(
    s1,
    language='en-us',
    backend='espeak',
    separator=Separator(phone=None, word=' ', syllable='|'),
    strip=True,
    preserve_punctuation=True)
phn
```

```python
phn = phonemize(
    s1,
    language='en-us',
    backend='espeak',
    separator=Separator(phone=None, word=' ', syllable='|'),
    strip=True,
    preserve_punctuation=False)
phn
```

```python
phn = phonemize(
    s1,
    language='en-us',
    backend='espeak',
    strip=True,
    preserve_punctuation=True)
phn
```

```python
phn = phonemize(
    " ".join(texts),
    language='en-us',
    backend='espeak',
    separator=Separator(phone=None, word=' ', syllable='|'),
    strip=True,
    preserve_punctuation=True)

texts, phn
```

```python
all_ph = list(set(phn))
print(all_ph)
```

```python
backend = EspeakBackend(
    language='en-us',
    preserve_punctuation=True,
    with_stress=True
)
```

```python
backend.phonemize(texts, strip=True)
```
