```python
from tts.tokenizer import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
disco_vocab = [
    "Existence is like a disco, a dance of light and shadow, of being and not being.",
    "This is a man with a lot of past, but little present. And almost no future.",
    "It's better to know you're being played than to be played without knowing it, is it not?",
    "In the dark times, should the stars also go out?",
    "Men without ideals are only animals."
]
```

```python
tokenizer.build_from_text(disco_vocab)
```

```python
tokenizer.vocab_path
```

```python
s = disco_vocab[1]
ids = tokenizer.encode(s)
print(ids)
```

```python
print(s)
for id in ids:
    print(tokenizer.id_to_token[id], end="")
```
